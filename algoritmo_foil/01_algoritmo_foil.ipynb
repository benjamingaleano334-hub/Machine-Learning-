
## 🚀 Ejercicio 1: Identificar empleados en formación

### 1. Separar positivos y negativos
- **Positivos (en_formacion = True):**
  - IT, terciario, edad 22
  - IT, universitario, edad 24
  - RRHH, terciario, edad 21
  - IT, terciario, edad 23

- **Negativos (en_formacion = False):**
  - IT, universitario, edad 35
  - Finanzas, maestría, edad 40
  - RRHH, universitario, edad 29
  - Finanzas, universitario, edad 38

### 2. Preguntas guía
- **Departamentos solo en positivos:** ninguno exclusivo (IT y RRHH aparecen en ambos).  
- **Nivel educativo exclusivo en positivos:** **terciario**.  
- **Edades solo en positivos:** 21, 22, 23, 24.

### 3. Regla inducida
```
SI nivel_educativo = 'terciario' O edad ≤ 23
ENTONCES en_formacion = True
```

👉 Esta regla cubre todos los positivos y excluye los negativos.

### Código en Python

````python
datos = [
 {"edad": 22, "departamento": "IT", "nivel_educativo": "terciario", "en_formacion": True},
 {"edad": 24, "departamento": "IT", "nivel_educativo": "universitario", "en_formacion": True},
 {"edad": 21, "departamento": "RRHH", "nivel_educativo": "terciario", "en_formacion": True},
 {"edad": 35, "departamento": "IT", "nivel_educativo": "universitario", "en_formacion": False},
 {"edad": 40, "departamento": "Finanzas", "nivel_educativo": "maestría", "en_formacion": False},
 {"edad": 29, "departamento": "RRHH", "nivel_educativo": "universitario", "en_formacion": False},
 {"edad": 23, "departamento": "IT", "nivel_educativo": "terciario", "en_formacion": True},
 {"edad": 38, "departamento": "Finanzas", "nivel_educativo": "universitario", "en_formacion": False}
]

positivos = [d for d in datos if d["en_formacion"]]
negativos = [d for d in datos if not d["en_formacion"]]

# Valores únicos
niveles_pos = {d["nivel_educativo"] for d in positivos}
niveles_neg = {d["nivel_educativo"] for d in negativos}
niveles_exclusivos = niveles_pos - niveles_neg

edades_pos = {d["edad"] for d in positivos}
edades_neg = {d["edad"] for d in negativos}
edades_exclusivas = edades_pos - edades_neg

print("Niveles exclusivos en positivos:", niveles_exclusivos)
print("Edades exclusivas en positivos:", edades_exclusivas)
````

## 📊 Ejercicio 2: FOIL Gain

### a) Programa en Python para condición `nivel_educativo == 'terciario'`

````python
import math

P, N = 4, 4   # antes
p, n = 3, 0   # después

frac_despues = p / (p + n)
frac_antes = P / (P + N)

foil_gain = p * (math.log2(frac_despues) - math.log2(frac_antes))

print("Condición: nivel_educativo == 'terciario'")
print("P =", P, "N =", N)
print("p =", p, "n =", n)
print("p/(p+n) =", round(frac_despues, 3))
print("P/(P+N) =", round(frac_antes, 3))
print("log2(p/(p+n)) =", round(math.log2(frac_despues), 3))
print("log2(P/(P+N)) =", round(math.log2(frac_antes), 3))
print("FOIL Gain =", round(foil_gain, 3))
````


### b) Cálculo manual
- \( \frac{p}{p+n} = \frac{3}{3+0} = 1.0 \)  
- \( \frac{P}{P+N} = \frac{4}{8} = 0.5 \)  
- \( \log_2(1.0) = 0 \)  
- \( \log_2(0.5) = -1 \)  
- FOIL Gain = \( 3 \cdot (0 - (-1)) = 3.0 \) ✅


Interpretación: la condición **nivel_educativo = terciario** es muy buena para separar positivos.

### c) Programa en Python para condición `edad ≤ 23`

````python
P, N = 4, 4
p, n = 3, 0

frac_despues = p / (p + n)
frac_antes = P / (P + N)

foil_gain = p * (math.log2(frac_despues) - math.log2(frac_antes))

print("Condición: edad ≤ 23")
print("P =", P, "N =", N)
print("p =", p, "n =", n)
print("p/(p+n) =", round(frac_despues, 3))
print("P/(P+N) =", round(frac_antes, 3))
print("log2(p/(p+n)) =", round(math.log2(frac_despues), 3))
print("log2(P/(P+N)) =", round(math.log2(frac_antes), 3))
print("FOIL Gain =", round(foil_gain, 3))
````


### d) Cálculo manual
- \( \frac{p}{p+n} = 1.0 \)  
- \( \frac{P}{P+N} = 0.5 \)  
- \( \log_2(1.0) = 0 \)  
- \( \log_2(0.5) = -1 \)  
- FOIL Gain = \( 3 \cdot (0 - (-1)) = 3.0 \) ✅


Interpretación: la condición **edad ≤ 23** también es muy buena para separar positivos.
