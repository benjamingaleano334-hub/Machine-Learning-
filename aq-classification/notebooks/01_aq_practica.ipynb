## 🚗 1) Clasificación de clientes que compran un automóvil eléctrico

### 🎯 Objetivo
Encontrar qué características tienen en común los clientes que compran un automóvil eléctrico y diferenciarlos de los que no lo compran.

### 📊 Datos
Ejemplos positivos y negativos:


```python
positivos = [
    {"edad": "Joven", "ingreso": "Alto", "tiene_garaje": "Si", "distancia_trabajo": "Corta"},
    {"edad": "Adulto", "ingreso": "Medio", "tiene_garaje": "Si", "distancia_trabajo": "Media"},
    {"edad": "Mayor", "ingreso": "Alto", "tiene_garaje": "Si", "distancia_trabajo": "Larga"}
]

negativos = [
    {"edad": "Joven", "ingreso": "Alto", "tiene_garaje": "No", "distancia_trabajo": "Corta"},
    {"edad": "Adulto", "ingreso": "Medio", "tiene_garaje": "No", "distancia_trabajo": "Media"},
    {"edad": "Mayor", "ingreso": "Bajo", "tiene_garaje": "No", "distancia_trabajo": "Larga"}]
```



### 🔎 Análisis
- En los positivos, todos tienen `tiene_garaje = Si`.  
- En los negativos, todos tienen `tiene_garaje = No`.  
- Los demás atributos varían y no son determinantes.  

👉 La característica clave es **tener garaje**.

### 🐍 Aplicación del algoritmo AQ

```python
atributos = positivos[0].keys()
reglas = []

for atributo in atributos:
    valores_positivos = {ejemplo[atributo] for ejemplo in positivos}
    valores_negativos = {ejemplo[atributo] for ejemplo in negativos}
    exclusivos = valores_positivos - valores_negativos
    if len(exclusivos) == 1:
        valor = list(exclusivos)[0]
        reglas.append((atributo, valor))

print("Características encontradas:")
for atributo, valor in reglas:
    print(atributo, "=", valor)
```


### ✅ Regla de clasificación

SI tiene_garaje = "Si"

ENTONCES Compra_Auto_Electrico = "Sí"


### 🧪 Verificación

```python
def clasificar_cliente(cliente):
    return "Sí" if cliente["tiene_garaje"] == "Si" else "No"

print("Ejemplos positivos:")
for cliente in positivos:
    print(clasificar_cliente(cliente))

print("\nEjemplos negativos:")
for cliente in negativos:
    print(clasificar_cliente(cliente))
```


👉 La regla cubre todos los positivos y excluye todos los negativos.

## 🏋️ 2) Inferir qué significa ser Socio Activo de un Gimnasio

### 🎯 Objetivo
Inducir la regla que diferencia a los socios activos de los no activos.

### 📊 Datos

```python
positivos = [
    {"edad": 25, "frecuencia": "frecuente", "plan": "premium"},
    {"edad": 32, "frecuencia": "frecuente", "plan": "premium"},
    {"edad": 45, "frecuencia": "frecuente", "plan": "premium"}
]

negativos = [
    {"edad": 25, "frecuencia": "ocasional", "plan": "básico"},
    {"edad": 32, "frecuencia": "ocasional", "plan": "estándar"},
    {"edad": 38, "frecuencia": "rara", "plan": "básico"}
]
```


### 🔎 Análisis
- Positivos: `frecuencia = frecuente` y `plan = premium`.  
- Negativos: frecuencias ocasional/rara y planes básicos/estándar.  
- La edad no es determinante.  

👉 La combinación clave es **frecuencia frecuente + plan premium**.

### 🐍 Aplicación del algoritmo AQ

```python
atributos = positivos[0].keys()
reglas = []

for atributo in atributos:
    valores_positivos = {ejemplo[atributo] for ejemplo in positivos}
    valores_negativos = {ejemplo[atributo] for ejemplo in negativos}
    exclusivos = valores_positivos - valores_negativos
    if len(exclusivos) == 1:
        valor = list(exclusivos)[0]
        reglas.append((atributo, valor))

print("Características encontradas:")
for atributo, valor in reglas:
    print(atributo, "=", valor)
```



### ✅ Regla de clasificación

SI frecuencia = "frecuente"

Y plan = "premium"

ENTONCES Socio_Activo = "Sí"

### 🧪 Verificación

```python
def clasificar_socio(socio):
    return "Sí" if socio["frecuencia"] == "frecuente" and socio["plan"] == "premium" else "No"

print("Ejemplos positivos:")
for socio in positivos:
    print(clasificar_socio(socio))

print("\nEjemplos negativos:")
for socio in negativos:
    print(clasificar_socio(socio))
```


👉 La regla cubre todos los positivos y excluye todos los negativos.



---

## 🎯 Conclusión general
- 🚗 Concesionaria: la condición clave es **tener garaje**.  
- 🏋️ Gimnasio: la condición clave es **frecuencia frecuente + plan premium**.  
- No siempre es necesario usar todos los atributos (ejemplo: la edad no fue útil en el gimnasio).  
- Las reglas inducidas clasifican correctamente todos los positivos y excluyen los negativos.
